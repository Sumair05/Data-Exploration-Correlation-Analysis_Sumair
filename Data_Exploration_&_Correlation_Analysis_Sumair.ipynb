```python
import pandas as pd
df=pd.read_csv('heart.csv')
df.head()
```

```python
import seaborn as sns
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10,12))
sns.heatmap(df.corr(),annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

```python
import seaborn as sns

sns.histplot(df['age'], kde=True)
plt.title('Age Distribution')
plt.show()
```

```python
sns.countplot(x='cp', data=df)
plt.title('Chest Pain Type Count')
plt.show()
```

```python
sns.histplot(df['age'], kde=True, bins=20)
plt.title('Age Distribution of Patients')
plt.show()
```

```python
sns.countplot(x='cp', data=df)
plt.title('Chest Pain Type Count')
plt.xlabel('Chest Pain Type')
plt.ylabel('Count')
plt.show()
```

### 🔍 Observations:

- The correlation matrix shows that the chest pain type (`cp`) has a strong positive correlation with heart disease (`target`).
- Maximum heart rate (`thalach`) is negatively correlated with heart disease, indicating lower heart rates in affected patients.
- Most patients are between 45 and 65 years old based on the age distribution.
- Chest pain type 0 (typical angina) is less common, while type 2 and 3 are more frequent and may indicate higher risk.
